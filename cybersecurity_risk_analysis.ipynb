```python
# Count failed login attempts for each user
failed_by_user = (
    df[df["Login_Status"] == "Failed"]
    .groupby("User_ID")
    .size()
)
```

```python
df["User_Failed_Login_Count"] = df["User_ID"].map(failed_by_user).fillna(0)
```

```python
df
```

```python
df["User_Failed_Login_Count"].describe()
```

```python
failed_by_ip = (
    df[df["Login_Status"] == "Failed"]
    .groupby("IP_Address")
    .size()
)
df["IP_Failed_Login_Count"] = (
    df["IP_Address"].map(failed_by_ip).fillna(0)
)
df["IP_Failed_Login_Count"].describe()
```

```python
users_per_ip = df.groupby("IP_Address")["User_ID"].nunique()
df["Users_Per_IP"] = df["IP_Address"].map(users_per_ip)
df["Users_Per_IP"].describe()
```

```python
locations_per_user = df.groupby("User_ID")["Location"].nunique()    
df["Locations_Per_User"] = df["User_ID"].map(locations_per_user)       
df["Locations_Per_User"].describe()
```

```python
df["Risk_Score"] = 0

# Rule 1: Failed login
df.loc[df["Login_Status"] == "Failed", "Risk_Score"] += 30

# Rule 2: Night login
df.loc[df["Is_Night_Login"] == "Yes", "Risk_Score"] += 10

# Rule 3: Repeated failed logins by user
df.loc[df["User_Failed_Login_Count"] >= 12, "Risk_Score"] += 15

# Rule 4: High failed login activity from IP
df.loc[df["IP_Failed_Login_Count"] >= 15, "Risk_Score"] += 20

# Rule 5: IP used by many users
df.loc[df["Users_Per_IP"] >= 5, "Risk_Score"] += 10

# Rule 6: User logging in from multiple locations
df.loc[df["Locations_Per_User"] >= 4, "Risk_Score"] += 10

# Rule 7: Locked account
df.loc[df["Account_Status"] == "Locked", "Risk_Score"] += 15





df["Risk_Score"].describe()     
df["Risk_Score"].value_counts().sort_index()
```

```python
def classify_risk(score):
    if score <= 20:
        return "Low"
    elif score <= 45:
        return "Medium"
    elif score <= 65:
        return "High"
    else:
        return "Critical"
df["Risk_Level"] = df["Risk_Score"].apply(classify_risk)
df["Risk_Level"].value_counts()
```
